```python
import requests

url = 'http://openapi.seoul.go.kr:8088/4768796e76646f6738376c4d4a6153/json/SebcCollegeInfoKor/1/100/'
r = requests.get(url)
```

```python
print(r.content)
```

```python
data = r.json()
data
```

```python
import pandas as pd
temp = []
for d in data['SebcCollegeInfoKor']['row']:
  temp.append(d)
uni = pd.DataFrame(temp)
uni.head()
```

```python
import requests

url = 'http://openapi.seoul.go.kr:8088/4768796e76646f6738376c4d4a6153/json/subwayStationMaster/1/1000/'
m = requests.get(url)
```

```python
print(m.content)
```

```python
data = m.json()
data
```

```python
import pandas as pd
temp = []
for d in data['subwayStationMaster']['row']:
  temp.append(d)
uni = pd.DataFrame(temp)
uni.head()
```

```python
import folium
anMap = folium.Map(location=[37.88988548,127.1371472],
                    zoom_control=False,
                    zoom_start=11,
                    control_scale= True)
anMap
```

```python

from folium.plugins import MarkerCluster
mc = MarkerCluster().add_to(anMap)

for n, row in uni.iterrows():
  folium.Marker(location=[row['LAT'], row['LOT']], popup = folium.Popup(row['BLDN_NM'], max_width=200)).add_to(mc)

anMap
```

```python
anMap.save('오픈API활용한folium맵표시.html')
```
